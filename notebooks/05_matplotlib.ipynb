```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.figure(figsize=(6,4))
x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x, y, label="Datos")
plt.title("Figura con plt.figure()")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, label="Datos")
ax.set_title("Figura con subplots")
ax.set_xlabel("Eje X")
ax.set_ylabel("Eje Y")
ax.legend()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

ax.plot(x, y)
ax.set_title("Uso de add_subplot")
plt.show()
```

```python
x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x,y)
plt.title("Scatter simple")
plt.show()
```

```python
sizes = np.random.rand(100)*100
colors = np.random.rand(100)

plt.scatter(x, y, s=sizes, c=colors)
plt.title("Scatter tamaño y color")
plt.show()
```

```python
coef = np.polyfit(x, y, 1)
line = np.poly1d(coef)

plt.scatter(x,y)
plt.plot(x, line(x))
plt.title("Scatter con regresión")
plt.show()
```

```python
x = np.arange(10)
y1 = x * 2
y2 = x * 3

plt.plot(x,y1)
plt.plot(x,y2)
plt.title("Comparación")
plt.show()
```

```python
y = x**2

coef = np.polyfit(x, y, 2)
poly = np.poly1d(coef)

plt.scatter(x,y)
plt.plot(x, poly(x))
plt.title("Regresión cuadrática")
plt.show()
```

```python
df = pd.DataFrame(np.random.rand(5,5), columns=list('ABCDE'))
corr = df.corr()

plt.imshow(corr, cmap='viridis')
plt.colorbar()
plt.title("Matriz de correlación")
plt.show()
```

```python
fig, ax = plt.subplots()
cax = ax.imshow(corr)

for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, round(corr.iloc[i,j],2),
                ha="center", va="center", color="white")

plt.title("Heatmap con valores")
plt.show()
```

```python
labels = ['A','B','C']
val1 = [10,20,15]
val2 = [5,10,8]

plt.bar(labels, val1)
plt.bar(labels, val2, bottom=val1)

plt.title("Barras apiladas")
plt.show()
```

```python
x = np.arange(len(labels))

plt.bar(x-0.2, val1, width=0.4)
plt.bar(x+0.2, val2, width=0.4)

plt.xticks(x, labels)
plt.title("Barras agrupadas")
plt.show()
```

```python
plt.bar(labels, val1, color='blue', hatch='/')
plt.bar(labels, val2, bottom=val1, color='orange', hatch='\\')

plt.title("Barras personalizadas")
plt.show()
```

```python
bars = plt.bar(labels, val1)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval,
             int(yval), ha='center', va='bottom')

plt.title("Barras con etiquetas")
plt.show()
```

¿Cuál gráfica fue más útil para interpretar datos y por qué?

La gráfica más útil fue el scatter plot (gráfico de dispersión), ya que permite visualizar de forma clara la relación entre dos variables. A través de este tipo de gráfica es posible identificar patrones, tendencias, agrupaciones e incluso valores atípicos.

Además, cuando se combina con una línea de regresión, facilita la comprensión de la correlación entre variables, lo cual resulta fundamental en el análisis de datos.

¿Qué diferencias notaste entre usar pyplot y la API orientada a objetos?

Se identificaron diferencias importantes entre ambas interfaces:

Pyplot:
Es más sencilla y rápida de usar.
Ideal para gráficos básicos o análisis rápidos.
Funciona de manera global, lo que puede dificultar el control en proyectos grandes.
API Orientada a Objetos (OO):
Ofrece mayor control sobre cada elemento de la gráfica.
Permite trabajar con múltiples gráficos dentro de una misma figura.
Es más organizada y escalable para proyectos complejos.

En conclusión, pyplot es útil para comenzar, pero la API OO es más adecuada para desarrollos profesionales.

¿Qué tipo de gráfica consideras más adecuada para tu proyecto personal?

Para un proyecto personal enfocado en análisis de datos (por ejemplo, análisis de videojuegos o comportamiento de usuarios), las gráficas más adecuadas serían:

Scatter plots: para analizar relaciones entre variables (ej: ventas vs puntuación).
Heatmaps: para visualizar correlaciones entre múltiples variables.
Gráficas de barras: para comparar categorías (ej: géneros de videojuegos más vendidos).

Estas visualizaciones permiten obtener insights claros y facilitar la toma de decisiones basada en datos.
